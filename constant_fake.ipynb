```python
from data_processing import prepare_constant_fake_data
from graphic import plot_classification_report, plot_dual_confusion_matrix
from ml_utils import train_and_evaluate
```

```python
# Sabit Sahte Dağılım: tüm sahte haber + 2 kat gerçek haber
train_df, test_df = prepare_constant_fake_data(
    "data/real-news.txt", "data/fake-news.txt", real_multiplier=2
)

print("Eğitim seti dağılımı:")
print(train_df["is_real"].value_counts())
print("\nTest seti dağılımı:")
print(test_df["is_real"].value_counts())
```

```python
# Naive Bayes
nb_fig_learning, nb_fig_roc, nb_train_report, nb_test_report = train_and_evaluate(
    train_df, test_df, "nb"
)
plot_classification_report("Naive Bayes — Sabit Sahte", nb_train_report, nb_test_report)
nb_fig_roc.show()
```

```python
# Random Forest
rf_fig_learning, rf_fig_roc, rf_train_report, rf_test_report = train_and_evaluate(
    train_df, test_df, "rf"
)
plot_classification_report("Random Forest — Sabit Sahte", rf_train_report, rf_test_report)
rf_fig_roc.show()
```

```python
# Logistic Regression
lr_fig_learning, lr_fig_roc, lr_train_report, lr_test_report = train_and_evaluate(
    train_df, test_df, "lr"
)
plot_classification_report("Lojistik Regresyon — Sabit Sahte", lr_train_report, lr_test_report)
lr_fig_roc.show()
```

```python
# Support Vector Machine
svm_fig_learning, svm_fig_roc, svm_train_report, svm_test_report = train_and_evaluate(
    train_df, test_df, "svm"
)
plot_classification_report("SVM — Sabit Sahte", svm_train_report, svm_test_report)
svm_fig_roc.show()
```

```python
# Şekil 1 — SVM ve LR Karmaşıklık Matrisleri (Sabit Sahte Senaryosu)
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression

svm_pipe = Pipeline([("tfidf", TfidfVectorizer()), ("clf", SVC(kernel="linear", probability=True, random_state=42))])
svm_pipe.fit(train_df["text"], train_df["is_real"])
svm_preds = svm_pipe.predict(test_df["text"])

lr_pipe = Pipeline([("tfidf", TfidfVectorizer()), ("clf", LogisticRegression(l1_ratio=0, random_state=42))])
lr_pipe.fit(train_df["text"], train_df["is_real"])
lr_preds = lr_pipe.predict(test_df["text"])

plot_dual_confusion_matrix(
    "Şekil 1 — Doğrusal Modeller Karmaşıklık Matrisleri (Sabit Sahte)",
    test_df["is_real"], svm_preds, "SVM",
    test_df["is_real"], lr_preds, "Lojistik Regresyon",
)
```
