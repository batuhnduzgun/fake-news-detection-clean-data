```python
import pandas as pd

# Read the TSV file
df = pd.read_csv("fake_news_dataset.tsv", sep="\t")

# Filter rows where the 'label' column is False
filtered_df = df[df["label"] == "False"]

fake_news = filtered_df["tweet"]

# Write the fake news to a text file
with open("fake_news.txt", "w", encoding="utf-8") as f:
    for news in fake_news:
        if news:
            news = news.replace("\n", " ").strip()
            f.write(news + "\n")
```
